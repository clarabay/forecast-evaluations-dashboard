```python
import pandas as pd
import numpy as np
import datetime
from epiweeks import Week
import epiweeks
import matplotlib.pyplot as plt
import matplotlib as mpl
from collections import defaultdict
import seaborn as sns
from pathlib import Path
import requests
import os
from datetime import date, timedelta
import warnings
import matplotlib.colors as mcolors
from itertools import combinations
from mpl_toolkits.axes_grid1 import make_axes_locatable
from matplotlib.colors import LinearSegmentedColormap

warnings.filterwarnings('ignore')
```

```python
df = pd.read_parquet('./precomputed/flusight_hosp_wis.parquet')
```

```python
df = df[(df.reference_date=='2026-10-31') & (df.model == 'MOBS-EpyStrain_Flu')]
df = df[['model','location', 'horizon', 'reference_date', 'target_end_date', 'wis_ratio']]
```

```python
dftest = pd.read_parquet('./scored_pipeline-202643_wis.parquet')
dftest['model'] = 'augment test run'
dftest = dftest[['model','location', 'horizon', 'reference_date', 'target_end_date', 'wis_ratio']]
```

```python
dfall = pd.concat([df, dftest])
```

```python
fig, ax = plt.subplots(1,1)

sns.boxplot(x='wis_ratio', y='model', data=dfall, showfliers=False,color = '#17B1BF',width=.4,)

for patch in ax.patches:
    r, gr, b, a = patch.get_facecolor()
    patch.set_facecolor((r, gr, b, .3))
    
sns.swarmplot(x='wis_ratio', y='model', data=dfall, color = '#17B1BF',size=3)

plt.text(3.5,-.1,dfall.groupby('model').wis_ratio.median())

plt.xlabel('WIS Ratio (compared to FluSight Baseline)')
plt.ylabel('')
```
